```python
import pandas as pd
import os

# Netflix y Prime: schema tipo A (un solo archivo)
for nombre in ["netflix", "prime"]:
    carpeta = f"kaggle_data/{nombre}"
    archivos = [f for f in os.listdir(carpeta) if f.endswith(".csv")]
    print(f"\n{'='*60}")
    print(f"📂 {nombre.upper()} — archivo: {archivos[0]}")
    df = pd.read_csv(os.path.join(carpeta, archivos[0]))
    print(f"Filas: {len(df)} | Columnas: {list(df.columns)}")
    print(df.head(2))

# Disney, Apple TV, HBO Max: schema tipo B (titles.csv + credits.csv)
for nombre in ["disney", "appletv", "hbomax"]:
    carpeta = f"kaggle_data/{nombre}"
    archivos = os.listdir(carpeta)
    print(f"\n{'='*60}")
    print(f"📂 {nombre.upper()} — archivos: {archivos}")
    df = pd.read_csv(os.path.join(carpeta, "titles.csv"))
    print(f"Filas: {len(df)} | Columnas: {list(df.columns)}")
    print(df.head(2))
```

```python
def parse_duration(row):
    """Convierte '90 min' -> 90, o '2 Seasons' -> 2"""
    dur = str(row["duration"])
    if "min" in dur:
        return int(dur.replace(" min", "").strip())
    elif "Season" in dur:
        return int(dur.split(" ")[0])
    return None

def process_familia_a(carpeta, nombre_archivo, plataforma):
    df = pd.read_csv(f"kaggle_data/{carpeta}/{nombre_archivo}")

    out = pd.DataFrame()
    out["show_id"] = carpeta + "_" + df["show_id"].astype(str)
    out["titulo"] = df["title"]
    out["tipo"] = df["type"]
    out["plataforma"] = plataforma
    out["director"] = df["director"].fillna("")
    out["elenco"] = df["cast"].fillna("")
    out["pais"] = df["country"].fillna("").apply(lambda x: x.split(",")[0].strip() if x else "")
    out["fecha_agregado"] = pd.to_datetime(df["date_added"], errors="coerce").dt.strftime("%Y-%m-%d")
    out["anio_lanzamiento"] = df["release_year"]
    out["clasificacion"] = df["rating"].fillna("")
    out["duracion"] = df.apply(parse_duration, axis=1)
    out["genero"] = df["listed_in"].fillna("").apply(lambda x: x.split(",")[0].strip() if x else "")
    out["descripcion"] = df["description"].fillna("")
    out["puntaje"] = None  # No disponible en este dataset
    return out

df_netflix = process_familia_a("netflix", "netflix_titles.csv", "Netflix")
df_prime = process_familia_a("prime", "amazon_prime_titles.csv", "Prime Video")

print(df_netflix.shape, df_prime.shape)
df_netflix.head(3)
```

```python
def get_director_cast(carpeta):
    """Lee credits.csv y arma un diccionario: id -> (director, elenco)"""
    credits = pd.read_csv(f"kaggle_data/{carpeta}/credits.csv")

    directores = credits[credits["role"] == "DIRECTOR"].groupby("id")["name"].apply(
        lambda x: ", ".join(x)
    )
    actores = credits[credits["role"] == "ACTOR"].groupby("id")["name"].apply(
        lambda x: ", ".join(x.head(4))  # primeros 4 actores
    )
    return directores, actores

def parse_lista(valor):
    """Convierte \"['drama', 'romance']\" -> 'drama' (primer elemento)"""
    try:
        lista = ast.literal_eval(valor)
        return lista[0] if lista else ""
    except (ValueError, SyntaxError):
        return ""

def process_familia_b(carpeta, plataforma):
    df = pd.read_csv(f"kaggle_data/{carpeta}/titles.csv")
    directores, actores = get_director_cast(carpeta)

    out = pd.DataFrame()
    out["show_id"] = carpeta + "_" + df["id"].astype(str)
    out["titulo"] = df["title"]
    out["tipo"] = df["type"].map({"MOVIE": "Movie", "SHOW": "TV Show"}).fillna(df["type"])
    out["plataforma"] = plataforma
    out["director"] = df["id"].map(directores).fillna("")
    out["elenco"] = df["id"].map(actores).fillna("")
    out["pais"] = df["production_countries"].fillna("[]").apply(parse_lista)
    # Fecha de agregado: no disponible en este dataset -> usamos 1 de enero del año de estreno
    out["fecha_agregado"] = df["release_year"].apply(
        lambda y: f"{int(y)}-01-01" if pd.notna(y) else None
    )
    out["anio_lanzamiento"] = df["release_year"]
    out["clasificacion"] = df["age_certification"].fillna("")
    out["duracion"] = df.apply(
        lambda r: r["seasons"] if r["type"] == "SHOW" and pd.notna(r["seasons"]) else r["runtime"],
        axis=1
    )
    out["genero"] = df["genres"].fillna("[]").apply(parse_lista)
    out["descripcion"] = df["description"].fillna("")
    out["puntaje"] = df["imdb_score"]
    return out

df_disney = process_familia_b("disney", "Disney+")
df_appletv = process_familia_b("appletv", "Apple TV+")
df_hbomax = process_familia_b("hbomax", "HBO Max")

print(df_disney.shape, df_appletv.shape, df_hbomax.shape)
df_disney.head(3)
```

```python
# Unificar las 5 plataformas en un solo DataFrame
df_full = pd.concat(
    [df_netflix, df_prime, df_disney, df_appletv, df_hbomax],
    ignore_index=True
)

print(f"Total de títulos unificados: {len(df_full)}")
print("\nDistribución por plataforma:")
print(df_full["plataforma"].value_counts())
print("\nDistribución por tipo:")
print(df_full["tipo"].value_counts())
print("\n¿Cuántos tienen puntaje?")
print(df_full["puntaje"].notna().sum(), "de", len(df_full))
```

```python
n_total = 800
frac = n_total / len(df_full)

df_sample = df_full.groupby("plataforma", group_keys=False).apply(
    lambda x: x.sample(frac=frac, random_state=42)
).reset_index(drop=True)

print(f"Muestra final: {len(df_sample)} títulos\n")
print("Distribución por plataforma:")
print(df_sample["plataforma"].value_counts())
print("\nDistribución por tipo:")
print(df_sample["tipo"].value_counts())
print("\n¿Cuántos tienen puntaje en la muestra?")
print(df_sample["puntaje"].notna().sum(), "de", len(df_sample))
```
